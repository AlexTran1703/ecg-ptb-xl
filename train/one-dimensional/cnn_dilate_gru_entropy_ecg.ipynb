```python
def normalize_data(X: np.ndarray) -> np.ndarray:
    X_min = np.min(X, axis=1, keepdims=True)
    X_max = np.max(X, axis=1, keepdims=True)
    X_normalized = (X - X_min) / (X_max - X_min + 1e-8)
    return X_normalized
```

```python
def normalize_data_z_score(data: np.ndarray) -> np.ndarray:
    num_samples, sample_length, num_channels = data.shape
    data_reshaped = data.reshape(-1, num_channels)
    scaler = StandardScaler()
    data_normalized = scaler.fit_transform(data_reshaped)
    return data_normalized.reshape(num_samples, sample_length, num_channels)
```

```python
import torch
import torch.nn as nn

class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_sizes=[3, 3], dilation_rates=[1, 2]):
        super().__init__()
        layers = []
        for i, (k, d) in enumerate(zip(kernel_sizes, dilation_rates)):
            layers.append(nn.Conv1d(
                in_channels if i == 0 else out_channels,
                out_channels,
                kernel_size=k,
                padding=(k // 2) * d,
                dilation=d
            ))
            layers.append(nn.BatchNorm1d(out_channels))
            layers.append(nn.LeakyReLU(0.01))
        self.block = nn.Sequential(*layers)

    def forward(self, x):
        return self.block(x)


class ResNetBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size=3, padding=1):
        super().__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=padding)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.relu = nn.LeakyReLU(0.01)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=padding)
        self.bn2 = nn.BatchNorm1d(out_channels)

        if in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv1d(in_channels, out_channels, kernel_size=1, bias=False),
                nn.BatchNorm1d(out_channels)
            )
        else:
            self.shortcut = nn.Identity()

    def forward(self, x):
        shortcut = self.shortcut(x)
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += shortcut
        return self.relu(out)


class MultiStageCNNGRU(nn.Module):
    def __init__(self, in_channels=12, num_classes=5, gru_hidden_size=64, gru_layers=1):
        super().__init__()

        # First CNN block with dilation
        self.cnn_block1 = CNNBlock(in_channels, 16, kernel_sizes=[5, 3], dilation_rates=[1, 2])

        self.gru1 = nn.GRU(
            input_size=16, hidden_size=gru_hidden_size, num_layers=gru_layers,
            batch_first=False, bidirectional=True
        )

        self.res_block1 = ResNetBlock(gru_hidden_size * 2, 64)

        # Second CNN block with larger receptive field
        self.cnn_block2 = CNNBlock(64, 16, kernel_sizes=[5, 5], dilation_rates=[2, 4])

        self.gru2 = nn.GRU(
            input_size=16, hidden_size=gru_hidden_size, num_layers=gru_layers,
            batch_first=False, bidirectional=True
        )

        self.cnn_block3 = CNNBlock(gru_hidden_size * 2, 64, kernel_sizes=[3, 3], dilation_rates=[4, 8])

        self.res_block2 = ResNetBlock(64, 64)

        # Final classifier
        self.global_pool = nn.AdaptiveAvgPool1d(1)
        self.classifier = nn.Sequential(
            nn.Flatten(),                # (B, C, 1) → (B, C)
            nn.Linear(64, 128),
            nn.LeakyReLU(0.01),
            nn.Dropout(0.3),
            nn.Linear(128, 64),
            nn.LeakyReLU(0.01),
            nn.Dropout(0.3),
            nn.Linear(64, num_classes)
        )

    def forward(self, x):
        # Input: (B, C=12, T)
        x = self.cnn_block1(x)  # → (B, 32, T)
        x = x.permute(2, 0, 1)  # → (T, B, C)
        x, _ = self.gru1(x)     # → (T, B, H*2)
        x = x.permute(1, 2, 0)  # → (B, H*2, T)

        x = self.res_block1(x)  # → (B, 64, T)

        x = self.cnn_block2(x)  # → (B, 64, T)
        x = x.permute(2, 0, 1)  # → (T, B, 64)
        x, _ = self.gru2(x)     # → (T, B, H*2)
        x = x.permute(1, 2, 0)  # → (B, H*2, T)

        x = self.cnn_block3(x)  # → (B, 64, T)
        x = self.res_block2(x)  # → (B, 64, T)

        x = self.global_pool(x)  # → (B, 64, 1)
        out = self.classifier(x)  # → (B, num_classes)
        return out
```

```python
import math
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

def approximate_entropy_gpu(u: torch.Tensor, m: int = 2, r: float = 0.2) -> torch.Tensor:
    """
    Batched Approximate Entropy computed on GPU.
    Args:
        u: Tensor of shape (B, T)
        m: embedding dimension
        r: tolerance (fraction of std)
    Returns:
        Tensor of shape (B,) - ApEn per sample
    """
    B, N = u.shape
    std = torch.std(u, dim=1, keepdim=True)  # (B, 1)
    r = r * std

    def _phi(m):
        # Create sequences of length m
        X = torch.stack([u[:, i:i + m] for i in range(N - m + 1)], dim=1)  # (B, N-m+1, m)
        # Compute Chebyshev distance: max(abs(x_i - x_j))
        X1 = X.unsqueeze(2)  # (B, N-m+1, 1, m)
        X2 = X.unsqueeze(1)  # (B, 1, N-m+1, m)
        D = torch.max(torch.abs(X1 - X2), dim=-1).values  # (B, N-m+1, N-m+1)
        C = (D <= r.unsqueeze(-1)).float().mean(dim=-1)  # (B, N-m+1)
        return torch.log(C + 1e-8).mean(dim=1)  # (B,)

    return (_phi(m) - _phi(m + 1)).abs()  # (B,)



class SincConv(nn.Module):
    def __init__(self, out_channels, kernel_size, sample_rate=1000, stride=1, padding='same',
                 min_low_hz=1.0, min_band_hz=2.0):
        super().__init__()
        self.out_channels = out_channels
        self.kernel_size = kernel_size
        self.sample_rate = sample_rate
        self.stride = stride
        self.min_low_hz = min_low_hz
        self.min_band_hz = min_band_hz

        # Initialize cutoff frequencies on Mel scale
        low_hz = min_low_hz
        high_hz = sample_rate / 2 - (min_low_hz + min_band_hz)
        mel = np.linspace(self.to_mel(low_hz), self.to_mel(high_hz), out_channels + 1)
        hz = self.to_hz(mel)

        self.low_hz_ = nn.Parameter(torch.Tensor(hz[:-1]))
        self.band_hz_ = nn.Parameter(torch.Tensor(np.diff(hz)))

        # Hamming window and sinc time axis
        n = torch.arange(kernel_size) - (kernel_size - 1) / 2
        self.register_buffer('window', 0.54 - 0.46 * torch.cos(2 * math.pi * n / kernel_size))
        self.register_buffer('n', 2 * math.pi * n / sample_rate)

        self.padding = kernel_size // 2 if padding == 'same' else 0

    @staticmethod
    def to_mel(hz):
        return 2595 * np.log10(1 + hz / 700)

    @staticmethod
    def to_hz(mel):
        return 700 * (10**(mel / 2595) - 1)

    def forward(self, x):
        low = self.min_low_hz + torch.abs(self.low_hz_)
        band = self.min_band_hz + torch.abs(self.band_hz_)
        high = torch.clamp(low + band, self.min_low_hz, self.sample_rate / 2)

        f_low = low[:, None]
        f_high = high[:, None]

        sinc_low = torch.sin(f_low * self.n) / (self.n + 1e-8)
        sinc_high = torch.sin(f_high * self.n) / (self.n + 1e-8)
        band_pass = (sinc_high - sinc_low) * self.window[None, :]
        band_pass = band_pass / (2 * band[:, None])

        filters = band_pass.view(self.out_channels, 1, self.kernel_size)
        return F.conv1d(x, filters, stride=self.stride, padding=self.padding)

class ApEnSincNet(nn.Module):
    def __init__(self, in_channels=1, num_classes=5, sample_rate=1000):
        super().__init__()
        self.sinc = SincConv(out_channels=32, kernel_size=251, sample_rate=sample_rate)
        self.bn = nn.BatchNorm1d(32)
        self.relu = nn.LeakyReLU(0.01)
        self.conv = nn.Conv1d(32, 16, kernel_size=5, padding=2)
        self.pool = nn.AdaptiveAvgPool1d(1)

        self.classifier = nn.Sequential(
            nn.Linear(16 + 1, 64),  # Include ApEn feature
            nn.LeakyReLU(0.01),
            nn.Dropout(0.3),
            nn.Linear(64, num_classes)
        )

    def forward(self, x, apen=None):
        # x: (B, 1, T)
        x = self.sinc(x)            # (B, 32, T1)
        x = self.relu(self.bn(x))
        x = self.conv(x)             # (B, 16, T2)
        x = self.pool(x).squeeze(-1) # (B, 16)

        if apen is None:
            # Estimate Approximate Entropy on the raw signal
            signal = x[:, 0, :] if x.ndim == 3 else x
            apen = approximate_entropy_gpu(signal).unsqueeze(1)

        x = torch.cat([x, apen], dim=1)  # (B, 17)
        return self.classifier(x)
```

```python
class LateFusionModel(nn.Module):
    def __init__(self, cnn_gru_model, apen_sinc_models, fusion='average'):
        """
        Args:
            cnn_gru_model: main model for 12-lead ECG
            apen_sinc_models: list of ApEnSincNet models (e.g., [II, V1, V6])
            fusion: 'average' or 'concat'
        """
        super().__init__()
        self.cnn_gru = cnn_gru_model
        self.apen_sinc_models = nn.ModuleList(apen_sinc_models)
        self.fusion = fusion

        if fusion == 'concat':
            total_dim = (len(apen_sinc_models) + 1) * 5  # Each model outputs 5 logits
            # self.fusion_layer = nn.Sequential(
            #     nn.Linear(total_dim, 5)
            # )
            self.fusion_layer = nn.Sequential(
                nn.Linear(total_dim, 5)
            )

    def forward(self, x12, apen_inputs, apen_feats=None):
        """
        Args:
            x12: Tensor (B, 12, T) - for cnn_gru
            apen_inputs: list of Tensors [(B, 1, T), (B, 1, T), ...] - one per ApEnSincNet
            apen_feats: list of optional ApEn features [(B, 1), ...] or None
        """
        out_main = self.cnn_gru(x12)  # (B, 5)
        out_apens = []

        for i, model in enumerate(self.apen_sinc_models):
            apen_feat_i = apen_feats[i] if apen_feats is not None else None
            out_apens.append(model(apen_inputs[i], apen_feat_i))  # each (B, 5)

        if self.fusion == 'average':
            return (out_main + sum(out_apens)) / (1 + len(out_apens))
        elif self.fusion == 'concat':
            all_outs = [out_main] + out_apens
            return self.fusion_layer(torch.cat(all_outs, dim=1))  # (B, 5)
        else:
            raise ValueError("Unknown fusion method")
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

selected_leads = [0, 1, 2, 6, 11]

apen_models = [ApEnSincNet(1, 5).to(device) for _ in range(len(selected_leads))]

model = LateFusionModel(
    cnn_gru_model=MultiStageCNNGRU(in_channels=12, num_classes=5).to(device),
    apen_sinc_models=apen_models,
    fusion='concat'
).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCEWithLogitsLoss()
```

```python
def extract_leads(inputs, lead_indices):
    # Returns transposed leads: [(B, 1, T), ...]
    return [inputs[:, :, i:i+1].transpose(1, 2) for i in lead_indices]

num_epochs = 50
best_val_loss = float('inf')
best_model_wts = None

for epoch in range(num_epochs):
    model.train()
    train_loss = 0.0

    for inputs, targets in train_loader:
        inputs, targets = inputs.to(device), targets.to(device)
        apen_inputs = extract_leads(inputs, selected_leads)
        x12 = inputs.transpose(1, 2)  # (B, T, 12)
        outputs = model(x12, apen_inputs)
        
        loss = criterion(outputs, targets)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        train_loss += loss.item() * inputs.size(0)

    train_loss /= len(train_loader.dataset)

    # Validation
    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for inputs, targets in val_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            apen_inputs = extract_leads(inputs, selected_leads)
            x12 = inputs.transpose(1, 2)
            outputs = model(x12, apen_inputs)
            val_loss += criterion(outputs, targets).item() * inputs.size(0)

    val_loss /= len(val_loader.dataset)
    
    
    # Print with color and progress bar
    epoch_str = f"\033[1;36mEpoch {epoch+1}/{num_epochs}\033[0m"
    bar_len = num_epochs
    filled_len = int(round(bar_len * (epoch+1) / float(num_epochs)))
    bar = '=' * filled_len + '-' * (bar_len - filled_len)
    print(f"{epoch_str} [{bar}] | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        print(f"Found best save with Val Loss: {val_loss:.4f}")
        best_model_wts = model.state_dict()  # Save the best model
        epochs_without_improvement = 0  # Reset the counter

model.eval()
test_loss = 0.0

with torch.no_grad():
    for inputs, targets in test_loader:
        inputs, targets = inputs.to(device), targets.to(device)
        leads = extract_leads(inputs, selected_leads)
        outputs = model(inputs.transpose(1, 2), leads)
        loss = criterion(outputs, targets)
        test_loss += loss.item() * inputs.size(0)

test_loss /= len(test_loader.dataset)
print(f"Test Loss: {test_loss:.4f}")
```

```python
def performance_evaluation(model=model, dataset = test_loader, dataset_name = "Test Set"):
    
    all_targets = []
    all_preds = []

    model.eval()
    with torch.no_grad():
        for inputs, targets in dataset:
            inputs, targets = inputs.to(device), targets.to(device)
            apen_inputs = extract_leads(inputs, selected_leads)
            outputs = model(inputs.transpose(1, 2), apen_inputs)
            probs = torch.sigmoid(outputs)
            preds = (probs > 0.5).int()

            all_targets.append(targets.cpu().numpy())
            all_preds.append(preds.cpu().numpy())

    all_targets = np.vstack(all_targets)
    all_preds = np.vstack(all_preds)

    num_classes = all_targets.shape[1]

    # Prepare table data
    table_data = []
    headers = ["Class", "Precision", "Recall (Sensitivity)", "Specificity", "F1-Score", "TN", "FP", "FN", "TP"]

    for i in range(num_classes):
        y_true = all_targets[:, i]
        y_pred = all_preds[:, i]

        precision = precision_score(y_true, y_pred, zero_division=0)
        recall = recall_score(y_true, y_pred, zero_division=0)
        cm = confusion_matrix(y_true, y_pred)

        if cm.size == 4:
            TN, FP, FN, TP = cm.ravel()
            total = TN + FP + FN + TP
            specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
            f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
            # Format with percentage
            TN_str = f"{TN} ({TN/total*100:.1f}%)"
            FP_str = f"{FP} ({FP/total*100:.1f}%)"
            FN_str = f"{FN} ({FN/total*100:.1f}%)"
            TP_str = f"{TP} ({TP/total*100:.1f}%)"
        else:
            TN_str = FP_str = FN_str = TP_str = "0 (0.0%)"
            specificity = 0.0

        table_data.append([
            i,
            f"{precision:.4f}",
            f"{recall:.4f}",
            f"{specificity:.4f}",
            f"{f1_score:.4f}",
            TN_str,
            FP_str,
            FN_str,
            TP_str,
        ])
    print(f"\n------------Performance Evaluation Metrics per Class({dataset_name}) -----------\n")
    print(tabulate(table_data, headers=headers, tablefmt="grid"))
    return all_targets, all_preds
```

```python
targets_train, preds_train = performance_evaluation(model, train_loader, "Train Set")
targets_val, preds_val = performance_evaluation(model, val_loader, "Val Set")
targets_test, preds_test = performance_evaluation(model, test_loader, "Test Set")
```

```python
def plot_confusion_matrix_per_class_multi(y_true_list, y_pred_list, class_names=None, set_names=None):
    """
    Plot confusion matrices for multiple sets (e.g., test, val, train) in columns.
    y_true_list, y_pred_list: list of arrays, each for a set.
    set_names: list of names for each set.
    """
    metric_label = ["TN", "FP", "FN", "TP"]
    num_sets = len(y_true_list)
    num_classes = y_true_list[0].shape[1]
    if set_names is None:
        set_names = [f"Set {i+1}" for i in range(num_sets)]
    if class_names is None:
        class_names = [str(i) for i in range(num_classes)]

    for class_idx in range(num_classes):
        fig, axes = plt.subplots(1, num_sets, figsize=(5*num_sets, 4))
        if num_sets == 1:
            axes = [axes]
        for set_idx in range(num_sets):
            y_true = y_true_list[set_idx][:, class_idx]
            y_pred = y_pred_list[set_idx][:, class_idx]
            cm = confusion_matrix(y_true, y_pred)
            total = np.sum(cm)
            if cm.size == 4:
                annot = np.empty_like(cm).astype(str)
                for r in range(2):
                    for c in range(2):
                        count = cm[r, c]
                        row_sum = np.sum(cm[r, :])
                        row_percent = 100 * count / row_sum if row_sum > 0 else 0
                        annot[r, c] = f"{count}\n({metric_label[r*2+c]}:{row_percent:.1f}%)"
            else:
                annot = cm
            sns.heatmap(cm, annot=annot, fmt='', cmap='Blues', cbar=False,
                        xticklabels=['Pred 0', 'Pred 1'], yticklabels=['True 0', 'True 1'],
                        ax=axes[set_idx])
            axes[set_idx].set_title(f"{set_names[set_idx]}")
            axes[set_idx].set_xlabel('Predicted label')
            axes[set_idx].set_ylabel('True label')
        plt.suptitle(f'Confusion Matrix - {class_names[class_idx]}', fontsize=16)
        plt.tight_layout(rect=[0, 0, 1, 0.95])
        plt.show()
```
